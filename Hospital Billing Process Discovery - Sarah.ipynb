```python
import pm4py
import os
from pm4py.visualization.petri_net import visualizer as pn_vis
from pm4py.visualization.dfg import visualizer as dfg_vis
from pm4py.visualization.heuristics_net import visualizer as hn_vis

# 1. Load your compressed event log
file_path = "/Users/sarahkhalid/Downloads/Hospital Billing - Event Log.xes.gz"
print(f"Loading log from {file_path}...")
log = pm4py.read_xes(file_path)

# Output directory pointing to your Documents folder
output_dir = "/Users/sarahkhalid/Documents/process_mining_exports"
os.makedirs(output_dir, exist_ok=True)

# 2. Fig 1: DFG Process Map
print("Generating DFG...")
dfg, start_activities, end_activities = pm4py.discover_dfg(log)
gviz_dfg = dfg_vis.apply(dfg, log=log, variant=dfg_vis.Variants.FREQUENCY)
dfg_path = os.path.join(output_dir, "Fig1_DFG.png")
dfg_vis.save(gviz_dfg, dfg_path)

# 3. Fig 2: BPMN Model
print("Generating BPMN...")
bpmn_model = pm4py.discover_bpmn_inductive(log)
bpmn_path = os.path.join(output_dir, "Fig2_BPMN.png")
pm4py.save_vis_bpmn(bpmn_model, bpmn_path)

# 4. Fig 3: Inductive Petri Net
print("Generating Petri Net...")
net, im, fm = pm4py.discover_petri_net_inductive(log)
gviz_pn = pn_vis.apply(net, im, fm)
pn_path = os.path.join(output_dir, "Fig3_PetriNet.png")
pn_vis.save(gviz_pn, pn_path)

# 5. Fig 4: Heuristics Net
print("Generating Heuristics Net...")
heurs_net = pm4py.discover_heuristics_net(log)
gviz_hn = hn_vis.apply(heurs_net)
hn_path = os.path.join(output_dir, "Fig4_HeuristicsMiner.png")
hn_vis.save(gviz_hn, hn_path)

print(f"✅ Done! Find your 4 loose files in: {output_dir}")
```
